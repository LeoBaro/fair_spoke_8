```python
input_dir = "out_20250404_154256"
```

```python
# read csv
import pandas as pd
df = pd.read_csv(input_dir + "/benchmark_results.csv")
# compute mean and std of took column, grouped by num_workers and batch_size
mean_took = df.groupby(["num_workers", "batch_size"])["took"].mean()
std_took = df.groupby(["num_workers", "batch_size"])["took"].std()
print(f"Mean took: {mean_took}, Std took: {std_took}")
```
