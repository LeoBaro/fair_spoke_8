```python
tagging_model = spacy.load("en_core_web_trf")
```

```python
captions = [
    "Hedgehog over the village of Shipkovo",
    "Evacuation Of A Wounded Soldier By An Tote Bag by Luc De Jaeger",
    "Fat Chants playing on Platform One",
    "Modernist-style three-bedroomed house in Pwllheli, Gwynedd, North Wales",
    "Police walking out of his patrol car to speak with a man at a street corner - stock footage",
    "Pair of Vintage French Rush Armchairs For Sale 3",
    "The top of Mount Ararat is covered with snow and clouds",
    "Top view of chocolate banana bread with four slices cut on cutting board",
    "A man puts his hand on your knee under the dinner table.",
    "A view of green protected by bunkers at Wilshire Country Club",
    "Columbia Women's Sandy River Breathable Cargo Short with UPF 30 Sun Protection"
]
```

```python
pos_strings = []
```

```python
for caption in captions:
    doc = tagging_model(caption)
    pos_tags = [token.pos_ for token in doc]
    sorted_unique_tags = sorted(set(pos_tags))
    pos_string = "_".join(sorted_unique_tags)
    pos_strings.append(pos_string)
```

```python
new_pos_tags = pos_strings + fra_pos_strings
```
